# Feature Exploration

Explore post features and their correlation with engagement rates.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Connect to database
import os
from sqlalchemy import create_engine

engine = create_engine(os.getenv('SYNC_DATABASE_URL', 'postgresql://postgres:postgres@localhost:5432/social_engine'))
```

```python
# Load posts with metrics
query = """
SELECT p.*, pm.likes, pm.comments, pm.shares, pm.saves, pm.reach, pm.engagement_rate,
       EXTRACT(hour FROM p.published_at) as hour_of_day,
       EXTRACT(dow FROM p.published_at) as day_of_week,
       LENGTH(p.content_text) as text_length,
       ARRAY_LENGTH(REGEXP_SPLIT_TO_ARRAY(p.content_text, '#'), 1) - 1 as hashtag_count
FROM posts p
JOIN post_metrics pm ON pm.post_id = p.id
WHERE p.status = 'published'
"""
df = pd.read_sql(query, engine)
print(f'Loaded {len(df)} post-metric records')
df.head()
```
